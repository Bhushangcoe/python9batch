```python
import pandas as pd
import matplotlib.pyplot as plt


def main():
    data = {
        "Name": ["Bhushan", "Sneha", "Raj", "Priya", "Yash", "Nikita"],
        "Department": ["IT", "HR", "Finance", "IT", "Sales", "IT"],
        "Salary": [45000, 52000, 48000, 60000, 43000, 55000],
    }

    df = pd.DataFrame(data)

    print("Employee Records:")
    print(df)

    average_salary = df["Salary"].mean()
    print(f"\nAverage Salary: {average_salary:.2f}")

    high_salary_employees = df[df["Salary"] >= 50000]
    print("\nEmployees with salary >= 50000:")
    print(high_salary_employees)

    department_total = df.groupby("Department")["Salary"].sum().sort_values(ascending=False)
    print("\nDepartment-wise Salary:")
    print(department_total)

    department_total.plot(kind="bar", color="skyblue", title="Total Salary by Department")
    plt.xlabel("Department")
    plt.ylabel("Salary")
    plt.tight_layout()
    plt.show()


if __name__ == "__main__":
    main()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Seed for reproducibility
np.random.seed(42)

# Generate mock data representing an index score for different UAE regions
dubai_scores = np.random.normal(loc=75, scale=8, size=150)
abu_dhabi_scores = np.random.normal(loc=78, scale=6, size=150)
sharjah_scores = np.random.normal(loc=65, scale=10, size=150)

# Set style
sns.set_theme(style="whitegrid")

# Create the KDE plot
plt.figure(figsize=(10, 6))
sns.kdeplot(data=dubai_scores, label="Dubai", fill=True, alpha=0.3, linewidth=2)
sns.kdeplot(data=abu_dhabi_scores, label="Abu Dhabi", fill=True, alpha=0.3, linewidth=2)
sns.kdeplot(data=sharjah_scores, label="Sharjah", fill=True, alpha=0.3, linewidth=2)

# Formatting the visualization
plt.title("Kernel Density Estimate (KDE) of Regional Performance Metrics in UAE", fontsize=14, pad=15)
plt.xlabel("Index Score", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.legend(title="Emirate")

# Display the plot
plt.show()
```
